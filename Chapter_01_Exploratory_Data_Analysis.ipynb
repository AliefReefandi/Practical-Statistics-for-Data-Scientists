### 4.1 Exploratory Data Analysis

Exploratory Data Analysis (EDA) adalah proses untuk
mengeksplorasi dan memahami karakteristik suatu dataset
sebelum dilakukan analisis atau pemodelan lebih lanjut.

EDA dapat dilakukan menggunakan dua pendekatan utama,
yaitu analisis numerik dan visualisasi.

Analisis numerik digunakan untuk memperoleh informasi
statistik seperti mean, median, standard deviation,
dan percentile.

Sementara itu, visualisasi digunakan untuk melihat
pola, distribusi, hubungan antarvariabel, serta kemungkinan
adanya nilai yang tidak biasa atau outlier.

Dengan melakukan EDA, kita dapat memperoleh pemahaman awal
mengenai data dan menentukan metode analisis yang sesuai.

### 4.2 Estimates of Location

Estimates of location digunakan untuk mengetahui nilai yang
dapat mewakili bagian tengah dari sebuah data.

Beberapa cara yang dibahas dalam bagian ini adalah mean,
weighted mean, median, dan trimmed mean. Setiap metode
memiliki cara perhitungan yang berbeda dan bisa digunakan
sesuai dengan kondisi data.

#### 4.2.1 Mean

Mean atau rata-rata adalah nilai yang didapat dengan
menjumlahkan semua data lalu membaginya dengan jumlah data.

Contohnya, jika terdapat data 10, 20, dan 30:

Mean = (10 + 20 + 30) / 3 = 20

Mean mudah digunakan untuk melihat nilai rata-rata,
tetapi hasilnya bisa terpengaruh oleh data yang nilainya
terlalu besar atau terlalu kecil.

```python
# Menghitung mean secara manual
data = [10, 20, 30]

mean = sum(data) / len(data)

print("Mean:", mean)
```

**Interpretasi:**

Hasil mean adalah 20. Artinya, nilai rata-rata dari data
10, 20, dan 30 adalah 20.

#### 4.2.2 Weighted Mean

Weighted mean atau rata-rata berbobot digunakan ketika
setiap data memiliki tingkat kepentingan yang berbeda.

Pada mean biasa, setiap data memiliki bobot yang sama.
Sedangkan pada weighted mean, data tertentu bisa memiliki
pengaruh yang lebih besar karena memiliki bobot yang lebih tinggi.

Contohnya, nilai tugas dan ujian bisa memiliki bobot yang
berbeda dalam menghitung nilai akhir.

```python
# Contoh weighted mean
nilai = [80, 90]
bobot = [0.4, 0.6]

weighted_mean = sum(n * b for n, b in zip(nilai, bobot))

print("Weighted Mean:", weighted_mean)
```

**Interpretasi:**

Hasil weighted mean adalah 86. Nilai 90 memiliki pengaruh
lebih besar karena bobotnya 60%, sedangkan nilai 80 memiliki
bobot 40%.

#### 4.2.3 Median

Median adalah nilai yang berada di tengah setelah data
diurutkan dari nilai terkecil sampai terbesar.

Contohnya:

10, 20, 30, 40, 50

Nilai tengahnya adalah 30, sehingga mediannya adalah 30.

Median cukup berguna ketika data memiliki nilai yang sangat
besar atau sangat kecil karena median tidak terlalu
terpengaruh oleh nilai tersebut.

```python
import numpy as np

data = [10, 20, 30, 40, 50]

median = np.median(data)

print("Median:", median)
```

**Interpretasi:**

Median dari data tersebut adalah 30. Nilai ini berada di
tengah setelah semua data diurutkan.

#### 4.2.4 Trimmed Mean

Trimmed mean adalah rata-rata yang dihitung setelah sebagian
nilai paling kecil dan paling besar dihapus.

Cara ini dapat digunakan ketika terdapat nilai yang terlalu
jauh dari data lainnya.

Contohnya, jika sebuah dataset memiliki beberapa nilai yang
sangat tinggi atau sangat rendah, nilai tersebut dapat
mempengaruhi mean. Trimmed mean mengurangi pengaruh tersebut
dengan membuang sebagian nilai di bagian ujung data.

```python
from scipy.stats import trim_mean

data = [10, 20, 30, 40, 50, 100]

trimmed_mean = trim_mean(data, proportiontocut=0.1)

print("Trimmed Mean:", trimmed_mean)
```

### 4.3 Estimates of Variability

Variability digunakan untuk melihat seberapa jauh data
menyebar dari nilai pusatnya.

Dua dataset bisa memiliki rata-rata yang sama, tetapi
penyebaran datanya bisa berbeda. Oleh karena itu, kita
perlu melihat ukuran penyebaran selain mean atau median.

Beberapa ukuran yang digunakan adalah variance, standard
deviation, mean absolute deviation, median absolute
deviation, range, percentile, dan interquartile range.



#### 4.3.1 Variance dan Standard Deviation

Variance digunakan untuk melihat seberapa jauh data
menyebar dari mean. Standard deviation merupakan akar
dari variance.

Standard deviation lebih mudah dipahami karena satuannya
sama dengan data aslinya. Kedua ukuran ini cukup sensitif
terhadap nilai yang terlalu jauh dari data lainnya.



```python
import pandas as pd

state = pd.read_csv('state.csv')

state.head()
```

#### 4.3.1 Variance dan Standard Deviation

Variance digunakan untuk melihat seberapa jauh data
menyebar dari mean. Standard deviation merupakan akar
dari variance.

Standard deviation lebih mudah dipahami karena satuannya
sama dengan data aslinya. Kedua ukuran ini cukup sensitif
terhadap nilai yang terlalu jauh dari data lainnya.

#### 4.3.2 Mean Absolute Deviation

Mean absolute deviation atau MAD melihat rata-rata jarak
setiap data dari nilai mean.

Nilai absolut digunakan supaya selisih positif dan negatif
tidak saling menghilangkan.

```python
mean_abs_deviation = (
    state['Population'] - state['Population'].mean()
).abs().mean()

print("Mean Absolute Deviation:", mean_abs_deviation)
```

**Interpretasi:**

Hasil ini menunjukkan rata-rata jarak populasi setiap
negara bagian terhadap nilai rata-ratanya.

#### 4.3.3 Median Absolute Deviation

Median absolute deviation menggunakan median sebagai
titik tengah untuk menghitung penyebaran data.

Ukuran ini lebih tahan terhadap nilai yang terlalu besar
atau terlalu kecil dibandingkan ukuran yang menggunakan
mean.


```python
from statsmodels.robust.scale import mad

median_abs_deviation = mad(state['Population'])

print("Median Absolute Deviation:", median_abs_deviation)
```

**Interpretasi:**

MAD menunjukkan penyebaran data berdasarkan median.
Karena menggunakan median, hasilnya tidak terlalu
dipengaruhi oleh nilai populasi yang sangat besar.

#### 4.3.4 Percentile

Percentile digunakan untuk mengetahui posisi suatu nilai
dalam kumpulan data.

Misalnya, percentile ke-50 sama dengan median. Percentile
juga bisa digunakan untuk melihat bagian bawah, tengah,
dan atas dari data.

```python
state['Murder.Rate'].quantile(
    [0.05, 0.25, 0.5, 0.75, 0.95]
)
```

**Interpretasi:**

Hasil percentile menunjukkan posisi Murder Rate pada
beberapa bagian data.

Percentile 50% merupakan median, sedangkan percentile
25% dan 75% dapat digunakan untuk melihat bagian tengah
data.

#### 4.3.5 Interquartile Range

Interquartile Range atau IQR adalah jarak antara percentile
75% dan percentile 25%.

IQR digunakan untuk melihat seberapa lebar penyebaran
bagian tengah dari data.

```python
iqr = (
    state['Population'].quantile(0.75)
    - state['Population'].quantile(0.25)
)

print("IQR:", iqr)
```

**Interpretasi:**

IQR menunjukkan jarak antara 25% data terbawah dan
25% data teratas, sehingga kita bisa melihat penyebaran
50% data yang berada di tengah.

### 4.4 Exploring Data Distribution

Distribusi data menunjukkan bagaimana nilai-nilai dalam
dataset tersebar.

Pada bagian ini kita akan menggunakan beberapa visualisasi,
yaitu boxplot, frequency table, histogram, dan density plot.

#### 4.4.1 Boxplot

Boxplot digunakan untuk melihat distribusi data secara
ringkas.

Boxplot dapat menunjukkan median, quartile, dan beberapa
nilai yang jauh dari data lainnya.

```python
ax = (state['Population'] / 1_000_000).plot.box()

ax.set_ylabel('Population (millions)')
```

**Interpretasi:**

Boxplot membantu melihat posisi median dan penyebaran
populasi negara bagian. Beberapa negara bagian dengan
populasi sangat tinggi terlihat sebagai nilai yang jauh
dari sebagian besar data.

#### 4.4.2 Frequency Table

Frequency table membagi data menjadi beberapa kelompok
atau interval kemudian menghitung jumlah data pada setiap
kelompok.

```python
binnedPopulation = pd.cut(state['Population'], 10)

binnedPopulation.value_counts().sort_index()
```

**Interpretasi:**

Tabel tersebut menunjukkan jumlah negara bagian yang
masuk ke setiap kelompok populasi.

Dari sini kita dapat melihat bahwa sebagian besar negara
bagian berada pada kelompok populasi yang relatif rendah,
sedangkan hanya beberapa negara bagian yang memiliki
populasi sangat besar.

#### 4.4.3 Histogram

Histogram menampilkan frequency table dalam bentuk grafik.

Sumbu x menunjukkan rentang nilai, sedangkan sumbu y
menunjukkan jumlah data pada setiap rentang.

```python
ax = (state['Population'] / 1_000_000).plot.hist(
    figsize=(4, 4)
)

ax.set_xlabel('Population (millions)')
```

**Interpretasi:**

Histogram membantu melihat bahwa distribusi populasi
tidak merata. Sebagian besar negara bagian memiliki
populasi yang lebih kecil, sedangkan hanya beberapa
negara bagian yang memiliki populasi sangat besar.

#### 4.4.4 Density Plot

Density plot mirip dengan histogram, tetapi distribusinya
ditampilkan dalam bentuk garis yang lebih halus.

Grafik ini membantu melihat pola umum dari distribusi data.

```python
ax = state['Murder.Rate'].plot.hist(
    density=True,
    xlim=[0, 12],
    bins=range(1, 12)
)

state['Murder.Rate'].plot.density(ax=ax)

ax.set_xlabel('Murder Rate (per 100,000)')
```

**Interpretasi:**

Density plot memperlihatkan bagian nilai Murder Rate
yang paling banyak muncul. Garis yang lebih tinggi
menunjukkan area dengan konsentrasi data yang lebih besar.

### 4.5 Exploring Binary and Categorical Data

Data kategorikal berisi data yang dibagi ke dalam beberapa
kategori.

Untuk melihat data seperti ini, kita bisa menghitung jumlah
atau persentase setiap kategori.

Beberapa cara yang digunakan adalah mode, expected value,
bar chart, dan pie chart.

#### 4.5.1 Mode

Mode adalah nilai atau kategori yang paling sering muncul
dalam sebuah dataset.

```python
data = ['A', 'B', 'B', 'C', 'B', 'A']

mode = pd.Series(data).mode()

print(mode)
```

**Interpretasi:**

Hasilnya menunjukkan bahwa kategori B adalah mode karena
muncul paling banyak dalam data.

#### 4.5.2 Bar Chart

Bar chart digunakan untuk membandingkan jumlah atau
proporsi dari beberapa kategori.

Setiap kategori ditampilkan sebagai sebuah batang.

```python
dfw = pd.DataFrame({
    'Delay': [23.02, 30.40, 4.03, 0.12, 42.43]
}, index=['Carrier', 'ATC', 'Weather', 'Security', 'Inbound'])

ax = dfw.transpose().plot.bar(
    figsize=(6, 4),
    legend=False
)

ax.set_xlabel('Cause of delay')
ax.set_ylabel('Count')
```

**Interpretasi:**

Grafik menunjukkan perbedaan persentase keterlambatan
berdasarkan penyebabnya.

Kategori dengan batang yang lebih tinggi memiliki
persentase keterlambatan yang lebih besar.

#### 4.5.3 Pie Chart

Pie chart menampilkan proporsi setiap kategori dalam
bentuk bagian dari lingkaran.

Grafik ini cocok digunakan ketika kita ingin melihat
perbandingan proporsi antar kategori.

```python
dfw['Delay'].plot.pie(
    autopct='%1.1f%%',
    figsize=(6, 6)
)
```

**Interpretasi:**

Setiap bagian lingkaran menunjukkan proporsi dari masing-
masing penyebab keterlambatan.

### 4.6 Correlation

Correlation digunakan untuk melihat hubungan antara
dua variabel.

Nilai correlation berada antara -1 sampai 1.

Nilai yang mendekati 1 menunjukkan hubungan positif,
sedangkan nilai yang mendekati -1 menunjukkan hubungan
negatif. Nilai yang mendekati 0 menunjukkan hubungan
linear yang lemah.

#### 4.6.1 Correlation Coefficient

Correlation coefficient digunakan untuk mengukur
seberapa kuat hubungan antara dua variabel numerik.

```python
state[['Population', 'Murder.Rate']].corr()
```

**Interpretasi:**

Hasil correlation menunjukkan hubungan antara Population
dan Murder Rate.

Nilai correlation harus dilihat bersama konteks datanya,
karena correlation hanya menunjukkan hubungan dan bukan
berarti satu variabel menyebabkan variabel lainnya.

#### 4.6.2 Scatterplot

Scatterplot digunakan untuk melihat hubungan antara
dua variabel numerik.

Setiap titik pada grafik mewakili satu data.

```python
ax = state.plot.scatter(
    x='Population',
    y='Murder.Rate'
)

ax.set_xlabel('Population')
ax.set_ylabel('Murder Rate')
```

**Interpretasi:**

Scatterplot membantu melihat apakah terdapat pola
hubungan antara populasi dan Murder Rate.

Jika titik-titik membentuk pola tertentu, hal tersebut
dapat menunjukkan adanya hubungan antara kedua variabel.

### 4.7 Exploring Two or More Variables

Kita juga dapat melihat hubungan antara lebih dari dua
variabel.

Untuk melakukan hal ini, kita dapat menggunakan beberapa
jenis visualisasi seperti contingency table, hexagonal
binning, contour plot, dan violin plot.

#### 4.7.1 Contingency Table

Contingency table digunakan untuk melihat hubungan
antara variabel kategorikal.

Tabel ini menghitung jumlah data berdasarkan kombinasi
dari dua kategori.

```python
data = pd.DataFrame({
    'Gender': ['L', 'L', 'P', 'P', 'L', 'P'],
    'Status': ['A', 'B', 'A', 'B', 'A', 'A']
})

pd.crosstab(data['Gender'], data['Status'])
```

**Interpretasi:**

Tabel menunjukkan jumlah data berdasarkan kombinasi
Gender dan Status.

Dengan tabel ini kita dapat melihat kategori mana
yang memiliki jumlah data lebih banyak.

#### 4.7.2 Hexagonal Binning

Hexagonal binning digunakan ketika jumlah titik pada
scatterplot sangat banyak.

Data dikelompokkan ke dalam bentuk hexagon sehingga
kepadatan data lebih mudah dilihat.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 4))

plt.hexbin(
    state['Population'],
    state['Murder.Rate'],
    gridsize=20
)

plt.xlabel('Population')
plt.ylabel('Murder Rate')
plt.show()
```

**Interpretasi:**

Hexagonal binning membantu melihat bagian grafik yang
memiliki konsentrasi data lebih tinggi.

Cara ini dapat lebih mudah dibaca dibandingkan scatterplot
ketika jumlah data sangat banyak.

#### 4.7.3 Contour Plot

Contour plot digunakan untuk melihat kepadatan data
dari dua variabel.

Garis pada grafik menunjukkan area dengan tingkat
kepadatan yang berbeda.

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.kdeplot(
    x=state['Population'],
    y=state['Murder.Rate'],
    fill=False
)

plt.xlabel('Population')
plt.ylabel('Murder Rate')
plt.show()
```

**Interpretasi:**

Garis pada contour plot menunjukkan area dengan
kepadatan data yang berbeda.

Area dengan garis yang lebih rapat menunjukkan perubahan
kepadatan yang lebih besar.

#### 4.7.4 Violin Plot

Violin plot merupakan pengembangan dari boxplot yang
juga menunjukkan bentuk distribusi data.

Grafik ini berguna untuk membandingkan distribusi
beberapa kelompok.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Contoh data sederhana untuk beberapa kelompok
data_violin = pd.DataFrame({
    'Kelompok': ['A', 'A', 'A', 'A', 'B', 'B', 'B', 'B'],
    'Nilai': [10, 12, 13, 15, 18, 20, 21, 25]
})

sns.violinplot(
    data=data_violin,
    x='Kelompok',
    y='Nilai'
)

plt.xlabel('Kelompok')
plt.ylabel('Nilai')
plt.show()
```

**Interpretasi:**

Violin plot memperlihatkan bentuk penyebaran nilai pada
masing-masing kelompok.

Bagian yang lebih lebar menunjukkan bahwa data lebih
banyak terkumpul pada rentang tersebut.
